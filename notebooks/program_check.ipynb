```python
import cvxpy as cp
import numpy as np

r = 20
np.random.seed(1)
A = np.hstack((np.random.randn(r, 1), np.ones([r, 1])))
c = A[:, 0:1]                                          # 修正3
b = (10.0 * np.random.randn() * c) + (0.5 * np.random.randn(r, 1))
b = b.flatten()                                        # 修正1

x = cp.Variable(2)
obj = cp.Minimize(cp.sum_squares(A @ x - b))        # 修正2
P = cp.Problem(obj)
P.solve(verbose=True)
print(x.value)
```

```python
import cvxpy as cp
import numpy as np
m, n, gam = 10, 5, 100.0
np.random.seed(2)
A = np.random.randn(m,n)
b = 100 * np.random.randn(m)
x, z = cp.Variable(n), cp.Variable(n)
obj = cp.Minimize( sum(cp.square(A@x-b)) \
                  + (gam * sum(z)) )
cons = [z >= x,
        z >= -x]
P = cp.Problem(obj, cons)
P.solve(verbose=True)
print(x.value)
```
